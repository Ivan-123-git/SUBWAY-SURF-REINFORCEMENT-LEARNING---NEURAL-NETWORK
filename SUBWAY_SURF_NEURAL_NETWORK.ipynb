 A NEURAL NETWORK THAT PLAYS A SUBWAY-SURF LIKE GAME IN THE TERMINAL AUTONOMOUSLY USING PYTORCH

```python
import random
import torch
import torch.nn as nn
import torch.optim as optim
from subless import SubwayEnv
from torch.utils.data import DataLoader
```

```python
class SubAI(nn.Module): #NEURAL NETWORK DECLARATION or INITIATION
  def __init__(self): #THis is a constructor  definntion of the neural network and and when the class is called an object is created using this function
    super(SubAI , self).__init__() # This is the super() built in function used in class construction to give you access to the methods and properties of the base class in this case nn.Module or the neural network module , this is where the layers are created if the function is called
    self.fc1=nn.Linear(6,64)
    self.relu = nn.ReLU()
    self.fc2=nn.Linear(64,32)
    self.fc3=nn.Linear(32,5)

  def forward(self,x):
    x=self.fc1(x)
    x=self.relu(x)
    x=self.fc2(x)
    x=self.relu(x)
    return self.fc3(x)
```

```python
# -- The Environment , the model and optimizer
env = SubwayEnv()
model = SubAI()
criterion = nn.MSELoss()                  # Measures prediction error
optimizer=optim.Adam(model.parameters(),lr = 0.001)
```

```python
episodes = 500
gamma = 0.95
epsilon = 1.0
epsilon_min = 0.05
epsilon_decay=0.995
print("AI training about to started")
```

```python
for episode in range(episodes):
    state = env.reset()
    done = False
    total_reward = 0

    while not done:
        state_tensor = torch.tensor(state, dtype=torch.float32)

        if random.random() < epsilon:
            action = random.randint(0, 4)  # Explore randomly
        else:
            with torch.no_grad():
                q_values = model(state_tensor)
                action = torch.argmax(q_values).item() # Exploit brain

        next_state, reward, done = env.step(action)
        total_reward += reward

        next_state_tensor = torch.tensor(next_state, dtype=torch.float32)

        with torch.no_grad():
            next_q_values = model(next_state_tensor)
            max_next_q = torch.max(next_q_values)
            target_q = reward + (0 if done else gamma * max_next_q)  # FIXED: changed + to *

        current_q_values = model(state_tensor)
        target_q_values = current_q_values.clone()

        target_q_values[action] = target_q  # FIXED: Added this crucial line back in!

        optimizer.zero_grad()
        loss = criterion(current_q_values, target_q_values)
        loss.backward()
        optimizer.step()

        state = next_state

    if epsilon > epsilon_min:
        epsilon *= epsilon_decay

    if (episode + 1) % 50 == 0:
        print(f"Episode [{episode+1}/{episodes}] | Score: {env.score} | Coins: {env.coins} | Epsilon: {epsilon:.2f}")

print("Training Complete")
```

Defining My neural network for running autonomously without training

```python
from sub_vis import SubwayEnv

class SubwayAI(nn.Module):
    def __init__(self):
        super(SubwayAI, self).__init__()
        self.fc1 = nn.Linear(6, 64)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(64, 32)  # Changed from 64 to 32
        self.fc3 = nn.Linear(32, 5)   # Changed from 64 to 32

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        x = self.relu(x)
        return self.fc3(x)

# 2. Initialize the environment and the model
env = SubwayEnv()
model = SubwayAI()

# 3. Load your trained model weights
try:
    model.load_state_dict(torch.load("subway_ai_model.pth"))
    print("✅ Model weights loaded successfully!")
except FileNotFoundError:
    print("⚠️ Warning: 'subway_ai_model.pth' not found. Playing with random untrained weights.")

model.eval()  # Set model to evaluation mode
time.sleep(1)

# 4. Clear screen and hide cursor for clean rendering
import sys
sys.stdout.write("\033[?25l\033[2J")
sys.stdout.flush()

try:
    # Play 3 test games autonomously
    for episode in range(3):
        state = env.reset()
        done = False

        while not done:
            # Render the high-resolution ANSI visual frame in your terminal/Colab
            env.render()

            # Convert state vector to PyTorch tensor
            state_tensor = torch.tensor(state, dtype=torch.float32)

            # Ask the AI for the best action
            with torch.no_grad():
                q_values = model(state_tensor)
                action = torch.argmax(q_values).item()  # Pick highest Q-value

            # Take a step in the visual game environment
            state, reward, done = env.step(action)

            # Control frame rate (~30 FPS so it's watchable)
            time.sleep(0.03)

        # Pause briefly between games
        time.sleep(1.5)

finally:
    env.close()
    print("\n🏁 Autonomous play session ended.")
```
